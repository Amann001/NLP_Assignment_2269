# Q1. Brown Corpus – Tokenization and Stop-word Removal

### Requirements
1. Load the Brown Corpus and select at least 20 sentences.
2. Perform sentence and word-level tokenization.
3. Convert tokens to lowercase.
4. Remove punctuation and English stop words.
5. Display original and cleaned tokens for at least five sentences.


```python
# ============================================================
# Q1(a): Load Brown Corpus and select at least 20 sentences
# ============================================================

# Brown.raw() gives us raw text. We use sent_tokenize() to
# explicitly perform sentence-level tokenization.
brown_raw_text = brown.raw()

brown_sentences = sent_tokenize(brown_raw_text)[:20]

print("Number of selected sentences:", len(brown_sentences))
print("\nFirst 5 selected sentences:\n")

for i, sentence in enumerate(brown_sentences[:5], start=1):
    print(f"{i}. {sentence}")
```

```python
# ============================================================
# Q1(b-d): Word tokenization, lowercasing, punctuation removal,
#          and stop-word removal
# ============================================================

english_stopwords = set(stopwords.words("english"))

def clean_tokens(sentence):
    # Word-level tokenization
    tokens = word_tokenize(sentence)

    # Convert to lowercase
    tokens = [token.lower() for token in tokens]

    # Keep only alphabetic tokens and remove stop words
    cleaned = [
        token for token in tokens
        if token.isalpha() and token not in english_stopwords
    ]

    return tokens, cleaned


q1_results = []

for sentence in brown_sentences:
    original_tokens, cleaned = clean_tokens(sentence)
    q1_results.append({
        "Original Sentence": sentence,
        "Word Tokens": original_tokens,
        "Cleaned Tokens": cleaned
    })

q1_df = pd.DataFrame(q1_results)

print("Sentence-level and word-level tokenization completed.")
display(q1_df.head(5))
```

```python
# ============================================================
# Q1(e): Display original and cleaned tokens for 5 sentences
# ============================================================

for i in range(5):
    print("=" * 90)
    print(f"Sentence {i+1}:")
    print(brown_sentences[i])

    original_tokens, cleaned_tokens = clean_tokens(brown_sentences[i])

    print("\nOriginal word tokens:")
    print(original_tokens)

    print("\nCleaned tokens:")
    print(cleaned_tokens)
    print()
```

## Q1 Observation

Sentence tokenization divides the text into individual sentences, while word tokenization divides each sentence into individual tokens. Lowercasing makes words consistent, and removing punctuation and stop words produces a cleaner representation suitable for further NLP analysis.


# Q2. Stemming and Lemmatization

### Requirements
1. Extract at least 30 alphabetic words from the Brown Corpus.
2. Apply Porter Stemmer.
3. Apply WordNet Lemmatizer.
4. Include words such as `playing`, `studies`, `running`, `better`, and `cars`.
5. Create a DataFrame containing original, stemmed, and lemmatized words.


```python
# ============================================================
# Q2(a): Extract at least 30 alphabetic words
# ============================================================

# Get alphabetic words from the Brown Corpus.
brown_alpha_words = [
    word.lower()
    for word in brown.words()
    if word.isalpha()
]

# Take the first 30 unique words for the main comparison.
selected_words = list(dict.fromkeys(brown_alpha_words))[:30]

# Required examples from the assignment.
required_words = ["playing", "studies", "running", "better", "cars"]

# Add any required example that is not already present.
for word in required_words:
    if word not in selected_words:
        selected_words.append(word)

print("Number of words used for comparison:", len(selected_words))
print(selected_words)
```

```python
# ============================================================
# Q2(b-d): Porter Stemming and WordNet Lemmatization
# ============================================================

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# POS mapping for the required examples makes the lemmatization
# more meaningful. WordNet uses:
# n = noun, v = verb, a = adjective.
special_pos = {
    "playing": "v",
    "studies": "v",
    "running": "v",
    "better": "a",
    "cars": "n"
}

comparison = []

for word in selected_words:
    stemmed_word = stemmer.stem(word)

    # Use the appropriate POS for the required examples.
    # For other words, WordNet's default noun POS is used.
    pos = special_pos.get(word, "n")
    lemma = lemmatizer.lemmatize(word, pos=pos)

    comparison.append({
        "Original Word": word,
        "Stemmed": stemmed_word,
        "Lemmatized": lemma
    })

q2_df = pd.DataFrame(comparison)

display(q2_df)
```

## Q2 Observation

**Stemming** cuts a word down to a root-like form and may produce a form that is not a valid English word. **Lemmatization** uses vocabulary and linguistic information to return a meaningful base word. Therefore, stemming is generally simpler and faster, while lemmatization is more linguistically informed.


# Q3. Basic Spelling Correction using the NLTK Words Corpus

### Requirements
1. Load the NLTK Words Corpus.
2. Create at least 10 intentionally misspelled words.
3. Use approximate string matching to find corrections.
4. Display misspelled words and suggestions.
5. Correct at least five example sentences.


```python
# ============================================================
# Q3(a): Load the NLTK Words Corpus
# ============================================================

# Build a lowercase vocabulary from the NLTK Words Corpus.
word_vocabulary = set(
    word.lower()
    for word in words.words()
    if word.isalpha()
)

print("Number of vocabulary words:", len(word_vocabulary))
print("Sample vocabulary words:", list(sorted(word_vocabulary))[:20])
```

```python
# ============================================================
# Q3(b): Create at least 10 intentionally misspelled words
# ============================================================

misspelled_words = [
    "recieve",
    "teh",
    "adress",
    "definately",
    "seperate",
    "occured",
    "becuase",
    "enviroment",
    "langauge",
    "wierd",
    "acheive",
    "studing"
]

print("Misspelled words:")
print(misspelled_words)
```

```python
# ============================================================
# Q3(c-d): Approximate string matching and suggestions
# ============================================================

def correct_word(word, vocabulary, cutoff=0.60):
    """
    Find the closest vocabulary word using approximate
    string matching with difflib.
    """
    word_lower = word.lower()

    # If it is already a valid word, return it unchanged.
    if word_lower in vocabulary:
        return word_lower

    # Restrict candidates to words with a similar length.
    # This makes matching faster and usually more sensible.
    candidates = [
        candidate for candidate in vocabulary
        if abs(len(candidate) - len(word_lower)) <= 2
    ]

    matches = difflib.get_close_matches(
        word_lower,
        candidates,
        n=1,
        cutoff=cutoff
    )

    return matches[0] if matches else "No suggestion"


suggestion_rows = []

for misspelled in misspelled_words:
    suggestion = correct_word(misspelled, word_vocabulary)
    suggestion_rows.append({
        "Misspelled Word": misspelled,
        "Suggested Correction": suggestion
    })

q3_suggestions_df = pd.DataFrame(suggestion_rows)

display(q3_suggestions_df)
```

```python
# ============================================================
# Q3(e): Correct at least five example sentences
# ============================================================

example_sentences = [
    "I definately enjoy studing natural langauge.",
    "Teh enviroment of the lab was very good.",
    "Please recieve the adress becuase I need it.",
    "The error occured because the files were seperate.",
    "It is wierd to recieve an acheive message."
]

def correct_sentence(sentence):
    tokens = word_tokenize(sentence)
    corrected_tokens = []

    for token in tokens:
        if token.isalpha():
            corrected = correct_word(token, word_vocabulary)

            # Preserve capitalization of the original token.
            if token[0].isupper() and corrected != "No suggestion":
                corrected = corrected.capitalize()

            corrected_tokens.append(corrected)
        else:
            corrected_tokens.append(token)

    # Join tokens while keeping punctuation formatting natural.
    result = " ".join(corrected_tokens)
    result = re.sub(r"\s+([,.!?;:])", r"\1", result)

    return result


corrected_rows = []

for sentence in example_sentences:
    corrected = correct_sentence(sentence)
    corrected_rows.append({
        "Original Sentence": sentence,
        "Corrected Sentence": corrected
    })

q3_sentences_df = pd.DataFrame(corrected_rows)

display(q3_sentences_df)
```

## Q3 Observation

Approximate string matching compares a misspelled word with words in a known vocabulary and selects a close candidate. The quality of the correction depends on the vocabulary and similarity threshold, so a spelling-correction system may still require contextual information for difficult cases.


# Q4. Unigram, Bigram and Trigram Analysis

### Requirements
1. Use a suitable subset of the Brown Corpus.
2. Retain lowercase alphabetic tokens.
3. Generate unigrams, bigrams and trigrams.
4. Calculate N-gram frequencies.
5. Display the 10 most frequent unigrams, bigrams and trigrams.
6. Explain how increasing N captures more contextual information.


```python
# ============================================================
# Q4(a-b): Select and preprocess a Brown Corpus subset
# ============================================================

# Use the first 10,000 Brown Corpus words as a suitable subset.
brown_subset = brown.words()[:10000]

# Retain only lowercase alphabetic tokens.
processed_tokens = [
    word.lower()
    for word in brown_subset
    if word.isalpha()
]

print("Number of processed tokens:", len(processed_tokens))
print("\nFirst 50 processed tokens:")
print(processed_tokens[:50])
```

```python
# ============================================================
# Q4(c-d): Generate N-grams and calculate frequencies
# ============================================================

# Generate unigrams, bigrams and trigrams.
unigram_list = list(ngrams(processed_tokens, 1))
bigram_list = list(ngrams(processed_tokens, 2))
trigram_list = list(ngrams(processed_tokens, 3))

# Count frequencies.
unigram_freq = Counter(unigram_list)
bigram_freq = Counter(bigram_list)
trigram_freq = Counter(trigram_list)

print("Total unigrams:", len(unigram_list))
print("Total bigrams:", len(bigram_list))
print("Total trigrams:", len(trigram_list))
```

```python
# ============================================================
# Q4(e): Display the 10 most frequent unigrams
# ============================================================

top_unigrams = unigram_freq.most_common(10)

unigram_df = pd.DataFrame(
    [(ngram[0][0], count) for ngram, count in top_unigrams],
    columns=["Unigram", "Frequency"]
)

display(unigram_df)
```

```python
# ============================================================
# Q4(e): Display the 10 most frequent bigrams
# ============================================================

top_bigrams = bigram_freq.most_common(10)

bigram_df = pd.DataFrame(
    [(" ".join(ngram), count) for ngram, count in top_bigrams],
    columns=["Bigram", "Frequency"]
)

display(bigram_df)
```

```python
# ============================================================
# Q4(e): Display the 10 most frequent trigrams
# ============================================================

top_trigrams = trigram_freq.most_common(10)

trigram_df = pd.DataFrame(
    [(" ".join(ngram), count) for ngram, count in top_trigrams],
    columns=["Trigram", "Frequency"]
)

display(trigram_df)
```

## Q4(f). How increasing N captures more contextual information

- **Unigram (N = 1):** considers one word at a time. It captures word frequency but very little context.
- **Bigram (N = 2):** considers two consecutive words, so it captures short-range relationships between words.
- **Trigram (N = 3):** considers three consecutive words, providing more information about local word order and context.

As **N increases**, an N-gram contains more contextual information. However, larger N-grams also become more specific and usually occur less frequently, which can make the model more sparse.
