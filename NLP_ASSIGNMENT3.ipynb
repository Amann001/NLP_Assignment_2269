```python
import nltk
import pandas as pd
import numpy as np
import math

from collections import Counter
from nltk.corpus import brown, gutenberg
from nltk.util import ngrams
```

## 1. N-gram Analysis Using the Brown Corpus

```python
nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
news_words = brown.words(categories='news')

print('Total words available in Brown News Corpus:', len(news_words))

tokens = [word.lower() for word in news_words if word.isalpha()]
tokens = tokens[:5000]

print('Number of tokens:', len(tokens))
print('First 20 tokens:')
print(tokens[:20])

total_tokens = len(tokens)
vocabulary = set(tokens)
vocabulary_size = len(vocabulary)

print('Total Tokens:', total_tokens)
print('Vocabulary Size:', vocabulary_size)
```

```python
unigrams = list(ngrams(tokens, 1))
bigrams = list(ngrams(tokens, 2))
trigrams = list(ngrams(tokens, 3))

print('First 10 Unigrams:')
print(unigrams[:10])

print('\nFirst 10 Bigrams:')
print(bigrams[:10])

print('\nFirst 10 Trigrams:')
print(trigrams[:10])
```

```python
unigram_freq = Counter(unigrams)
bigram_freq = Counter(bigrams)
trigram_freq = Counter(trigrams)

print('Top 10 Unigrams:')
for item, count in unigram_freq.most_common(10):
    print(item, ':', count)

print('\nTop 10 Bigrams:')
for item, count in bigram_freq.most_common(10):
    print(item, ':', count)

print('\nTop 10 Trigrams:')
for item, count in trigram_freq.most_common(10):
    print(item, ':', count)
```

## 2. Bigram Probability and Sentence Scoring Using Jane Austen's Emma

```python
nltk.download('gutenberg')

words = gutenberg.words('austen-emma.txt')

print('Total words:', len(words))
print(words[:20])

gutenberg_tokens = [
    word.lower()
    for word in words
    if word.isalpha()
]

print('Number of normalized tokens:', len(gutenberg_tokens))
print(gutenberg_tokens[:20])
```

```python
unigram_counts = Counter(gutenberg_tokens)

bigram_list = list(ngrams(gutenberg_tokens, 2))
bigram_counts = Counter(bigram_list)

print('Vocabulary Size:', len(unigram_counts))
print('\nTop 10 Unigrams:')
print(unigram_counts.most_common(10))

print('\nTop 10 Bigrams:')
print(bigram_counts.most_common(10))

vocab_size = len(unigram_counts)

def bigram_probability(previous_word, current_word):
    bigram_count = bigram_counts[(previous_word, current_word)]
    previous_count = unigram_counts[previous_word]

    probability = (bigram_count + 1) / (previous_count + vocab_size)
    return probability

print('P(emma | the) =', bigram_probability('the', 'emma'))
```

```python
def sentence_score(sentence):
    words = [
        word.lower()
        for word in sentence.split()
        if word.isalpha()
    ]

    if len(words) < 2:
        return 0

    log_probability = 0

    for i in range(1, len(words)):
        probability = bigram_probability(words[i - 1], words[i])
        log_probability += math.log(probability)

    return math.exp(log_probability)

test_sentences = [
    'Emma was very happy',
    'She went to the house',
    'The weather was beautiful'
]

for sentence in test_sentences:
    score = sentence_score(sentence)
    print(f'{sentence} -> {score}')
```

## 3. Hidden Markov Model for POS Tagging

```python
nltk.download('universal_tagset')

tagged_sentences = brown.tagged_sents(
    categories='news',
    tagset='universal'
)

print('Number of tagged sentences:', len(tagged_sentences))
print(tagged_sentences[0])

split = int(0.8 * len(tagged_sentences))

train_data = tagged_sentences[:split]
test_data = tagged_sentences[split:]

print('Training sentences:', len(train_data))
print('Testing sentences:', len(test_data))
```

```python
from nltk.tag import HiddenMarkovModelTrainer

trainer = HiddenMarkovModelTrainer()
hmm_tagger = trainer.train_supervised(train_data)

print('HMM model trained successfully.')

accuracy = hmm_tagger.accuracy(test_data)

print('HMM POS Tagging Accuracy:', accuracy)
print('Accuracy: {:.2f}%'.format(accuracy * 100))
```

```python
test_sentences = [
    'The government announced a new policy',
    'The company reported strong profits',
    'The scientist discovered a new method',
    'The students completed their assignment',
    'The market showed significant growth'
]

for sentence in test_sentences:
    tokens = sentence.split()
    tagged = hmm_tagger.tag(tokens)

    print('\nSentence:', sentence)
    print('Tags:', tagged)
```

## 4. Term-Frequency Analysis Using the 20 Newsgroups Dataset

```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer

categories = ['sci.space', 'comp.graphics']

newsgroups = fetch_20newsgroups(
    subset='train',
    categories=categories,
    remove=('headers', 'footers', 'quotes')
)

print('Number of documents:', len(newsgroups.data))
print('Categories:', newsgroups.target_names)

vectorizer = CountVectorizer(
    stop_words='english',
    max_features=1000
)

X = vectorizer.fit_transform(newsgroups.data)
terms = vectorizer.get_feature_names_out()
term_counts = np.asarray(X.sum(axis=0)).flatten()

term_frequency = pd.DataFrame({
    'Term': terms,
    'Frequency': term_counts
})

term_frequency = term_frequency.sort_values(
    by='Frequency',
    ascending=False
)

print(term_frequency.head(20))
```

```python
graphics_data = fetch_20newsgroups(
    subset='train',
    categories=['comp.graphics'],
    remove=('headers', 'footers', 'quotes')
)

vectorizer_graphics = CountVectorizer(
    stop_words='english'
)

X_graphics = vectorizer_graphics.fit_transform(graphics_data.data)

graphics_terms = vectorizer_graphics.get_feature_names_out()
graphics_counts = np.asarray(X_graphics.sum(axis=0)).flatten()

graphics_df = pd.DataFrame({
    'Term': graphics_terms,
    'Frequency': graphics_counts
}).sort_values(
    by='Frequency',
    ascending=False
)

print('Top 20 COMP.GRAPHICS terms:')
print(graphics_df.head(20))
```
